```python
#Now Splitting the dataset in X_Train , Y_train
from sklearn.model_selection import train_test_split
```

```python
X_train ,X_test,y_train,y_test = train_test_split(x,y,random_state=14)
```

```python
#Import the genesim model for creating the embeddings
```

```python
import gensim
```

```python
from gensim.models import word2vec
```

```python
import gensim.downloader as api
```

```python
from nltk.tokenize import word_tokenize
```

```python
glove_model = api.load('glove-wiki-gigaword-50')
```

```python
import nltk
```

```python
nltk.download('punkt_tab')
```

```python
#Creating a list of tokens for creating embedding
#Here we had just two classes so we did not hot encoded this , if more than two classes than we need to one hot encoded
tokens=[]
for text in X_train:
    words = word_tokenize(text)
    words = [w.lower() for w in words if w.isalpha()]
    tokens.append(words)
    
```

```python
#Now we write function for creating the document vector
def document_vector(doc):
    doc = [word for word in doc if word in glove_model]
    if len(doc)==0:
        return np.zeros(glove_model.vector_size)    
    return np.mean(glove_model[doc],axis=0)
```

```python
sms_vecs = np.array([document_vector(doc) for doc in tokens])
```

```python
#Repeating the same steps for X_test
tokens_test=[]
for text in X_test:
    words = word_tokenize(text)
    words = [w.lower() for w in words if w.isalpha()]
    tokens_test.append(words)
```

```python
sms_vecs_test = np.array([document_vector(doc) for doc in tokens_test])
```
